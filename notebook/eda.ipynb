```python
import pandas as pd

df = pd.read_csv("../data/PCOS_data_without_infertility.csv")

print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_")
```

```python
df.columns
```

```python
df = df.drop(columns=['Sl._No', 'Patient_File_No.', 'Unnamed:_41'], errors='ignore')
```

```python
df.rename(columns={
    'PCOS_(Y/N)': 'PCOS (Y/N)',
    'Age_(yrs)': 'Age (yrs)',
    'Weight_(Kg)': 'Weight (Kg)',
    'Pulse_rate(bpm)': 'Pulse rate(bpm)',
    'RR_(breaths/min)': 'RR (breaths/min)',
    'Cycle_length(days)': 'Cycle length(days)',
    'Marraige_Status_(Yrs)': 'Marraige Status (Yrs)',
    'No._of_aborptions': 'No. of aborptions',
    'I___beta-HCG(mIU/mL)': 'I beta-HCG(mIU/mL)',
    'II____beta-HCG(mIU/mL)': 'II beta-HCG(mIU/mL)',
    'Waist:Hip_Ratio': 'Waist:Hip Ratio',
    'TSH_(mIU/L)': 'TSH (mIU/L)',
    'Vit_D3_(ng/mL)': 'Vit D3 (ng/mL)',
    'Weight_gain(Y/N)': 'Weight gain(Y/N)',
    'Skin_darkening_(Y/N)': 'Skin darkening (Y/N)',
    'Hair_loss(Y/N)': 'Hair loss(Y/N)',
    'Fast_food_(Y/N)': 'Fast food (Y/N)',
    'BP__Systolic_(mmHg)': 'BP _Systolic (mmHg)',
    'BP__Diastolic_(mmHg)': 'BP _Diastolic (mmHg)',
    'Follicle_No._(L)': 'Follicle No. (L)',
    'Follicle_No._(R)': 'Follicle No. (R)',
    'Avg._F_size_(L)_(mm)': 'Avg. F size (L) (mm)',
    'Avg._F_size_(R)_(mm)': 'Avg. F size (R) (mm)',
    'Endometrium_(mm)': 'Endometrium (mm)'
}, inplace=True)
```

```python
df.rename(columns={
    'Blood_Group': 'Blood Group'
}, inplace=True)
```

```python
print(df.columns.tolist())
```

```python
df.to_csv("../data/cleaned_pcos.csv", index=False)
```

```python
df.dtypes
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
sns.countplot(x='PCOS (Y/N)', data=df)
plt.title('PCOS Distribution')
plt.savefig('../reports/pcos_distribution.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age (yrs)'], bins=20, kde=True)
plt.title('Age Distribution')
plt.savefig('../reports/age_distribution.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='PCOS (Y/N)', y='BMI', data=df)
plt.title('BMI vs PCOS')
plt.savefig('../reports/bmi_vs_pcos.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='PCOS (Y/N)', y='Weight (Kg)', data=df)
plt.title('Weight vs PCOS')
plt.savefig('../reports/weight_vs_pcos.png', bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(df.corr(numeric_only=True),
            cmap='coolwarm',
            center=0)

plt.title('Feature Correlation Heatmap')
plt.savefig('../reports/correlation_heatmap.png',
            bbox_inches='tight')
plt.show()
```

```python
corr = df.corr(numeric_only=True)['PCOS (Y/N)']
corr = corr.abs().sort_values(ascending=False)[1:11]

plt.figure(figsize=(8,5))
corr.plot(kind='bar')
plt.title('Top Features Correlated with PCOS')
plt.ylabel('Absolute Correlation')
plt.tight_layout()

plt.savefig('../reports/top_features.png')
plt.show()
```
